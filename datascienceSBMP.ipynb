IMPORTING REQUIRED LIBRARIES

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score




```

LOAD DATASET

```python
df = pd.read_csv('archive.zip')
print("Dataset loaded successfully.\n")
```

DATA CLEANING AND PREPROCESSING


```python
# Drop the ID column, as it's not a feature
df = df.drop('customerID', axis=1)

# 'TotalCharges' should be a number. 'coerce' turns errors (like ' ') into 'NaN'
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Convert the target 'Churn' to a number (0 or 1)
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

df = df.dropna()

# --- 3. Feature Preprocessing (Simplified) ---

# Separate features (X) from the target (y)
X = df.drop('Churn', axis=1)
y = df['Churn']

# This finds all text-based columns and converts them into 0/1 columns.
X_processed = pd.get_dummies(X, drop_first=True)

# We list them so we only scale those and not the new 0/1 columns
numeric_features = ['tenure', 'MonthlyCharges', 'TotalCharges']

scaler = StandardScaler()

# Fit and transform only the numeric columns, updating them in our processed DataFrame
X_processed[numeric_features] = scaler.fit_transform(X_processed[numeric_features])

print("Data preprocessing complete.\n")
```

TRAINING AND TESTING

```python
X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42, stratify=y)

# Create the model
model = LogisticRegression(random_state=42, max_iter=1000)

# Train the model on the training data
print("Training the model...")
model.fit(X_train, y_train)
print("Model training complete.\n")

# --- 5. Evaluate the Model ---
print("--- Model Evaluation ---")
y_pred = model.predict(X_test)
```

RESULT

```python
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Not Churned (0)', 'Churned (1)']))
```
